## 8.Sorting & Searching

```python
arr = np.array([1,2,3,4,5,6,])
arr
```

```python
np.sort(arr)
#np.sort() return a sorted copy
```

```python
np.argsort(arr)
#indices that would sort the array
```

```python
np.searchsorted(np.sort(arr),10)
#.searchsorted ---index to insert a value and keep order
```

```python
np.nonzero(arr)
#indices of non zero elemnts
```

```python
np.unique(arr,return_counts=True)
#
```
